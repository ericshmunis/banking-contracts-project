```python
%sql
CREATE OR REPLACE FUNCTION banking_contracts_demo.bronze.mask_ssn(ssn STRING)
RETURNS STRING
RETURN CASE
  WHEN is_account_group_member('compliance_team') THEN ssn
  ELSE CONCAT('***-**-', RIGHT(ssn, 4))
END;
```

```python
%sql
ALTER TABLE banking_contracts_demo.bronze.accounts_daily
ALTER COLUMN customer_ssn SET MASK banking_contracts_demo.bronze.mask_ssn;
```

```python
%sql
SELECT account_id, customer_ssn, customer_name FROM banking_contracts_demo.bronze.accounts_daily LIMIT 5;
```
